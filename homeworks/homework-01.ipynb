```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(3)
```

```python
num_draws = 1000

draws = rng.integers(1, 7, size=(num_draws,2))
Y = draws[:, 0] + draws[:, 1]
X = np.maximum(draws[:, 0], draws[:, 1])
```

```python
model = LinearRegression()
model.fit(X.reshape(-1, 1), Y)

u = np.linspace(1, 6, 100).reshape(-1, 1)
v = model.predict(u)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(u, v, label="Y(0)", color="red", linewidth=2)
ax.scatter(X, Y, c='lightblue', s=1)
ax.set_title("sample full distribution")
ax.set_xlabel("X")
ax.set_ylabel("Y")
plt.show()
```

Problem 3 (1.22 KNN)

Sixteen batches of the plastic were made, and from each batch one test item was molded. Each test item was randomly assigned to one of the four predetermined time levels, and the hardness was measured after the assigned elapsed time. The results are shown below; X is the elapsed time in hours, and Y is hardness in Brinell units. Assume the first-order regression model (1.1) is appropriate (model (2.1) in the notes).

```python
y = np.array([199.0, 205.0, 196.0, 200.0, 218.0, 220.0, 215.0, 223.0, 237.0, 234.0, 235.0, 230.0, 250.0, 248.0, 253.0, 246.0])
x = np.array([16, 16, 16, 16, 24, 24, 24, 24, 32, 32, 32, 32, 40, 40, 40, 40])


ols_model = LinearRegression()
ols_model.fit(x.reshape(-1, 1), y)

a = np.linspace(16, 40, 100).reshape(-1, 1)
b = ols_model.predict(a)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(a, b, label="Y(0)", color="red", linewidth=2)
ax.scatter(x, y, c='lightblue', s=1)
ax.set_title("sample full distribution")
ax.set_xlabel("X")
ax.set_ylabel("Y")
plt.show()
```

```python
y_pred = ols_model.predict(x.reshape(-1, 1))
mse = mean_squared_error(y, y_pred)
mse
```
