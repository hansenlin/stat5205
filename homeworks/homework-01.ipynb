```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(3)
```

```python
num_draws = 1000

draws = rng.integers(1, 7, size=(num_draws,2))
Y = draws[:, 0] + draws[:, 1]
X = np.maximum(draws[:, 0], draws[:, 1])

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(X, Y, c='lightblue', s=1)
ax.set_title("sample full distribution")
ax.set_xlabel("X")
ax.set_ylabel("Y")
plt.show()
```
